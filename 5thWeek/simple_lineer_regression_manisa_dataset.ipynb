```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_path = '../Data/data.csv'
df = pd.read_csv(data_path)
df.head()
```

```python
df = df[[ 'area_mSqr' , 'list-view-price' ]]
# Change column name to turkish
df = df.rename(columns={'list-view-price': 'fiyat', 'area_mSqr': 'm2'})
df.head()
```

```python
x = df['m2'].values
y = df['fiyat'].values
x_line = np.linspace(x.min(), x.max(), 500)
```

```python
def sketch_regression(degree):
    coeff = np.polyfit(x, y, degree)
    model = np.poly1d(coeff)
    plt.figure(figsize=(10, 6))
    plt.scatter(x, y,color = 'blue', label='Reel Data')
    plt.plot(x_line, model(x_line), color='yellow', label=f'{degree}th Degree Regression Line')

    plt.ylim(y.min(), y.max())

    plt.title(f'{degree}th Degree Regression Line')
    plt.xlabel('m2')
    plt.ylabel('fiyat')
    plt.legend()
    plt.show()
```

```python
for degree in range(1, 27):
    sketch_regression(degree)
```
